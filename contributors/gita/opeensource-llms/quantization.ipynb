```python
import numpy as np
```

```python
w= np.array([0.12, -0.45, 0.03, 0.31, -0.08, 0.50, -0.27, 0.19])
```

```python
w.shape
```

```python
bits=4
block=4
```

```python
qmax=2**(bits-1)-1
```

```python
qmax
```

```python
shape=w.shape
```

```python
shape
```

```python
# reshape
w=w.reshape(-1, block) if block else w.reshape(1, -1)
w
```

Each row is one block, and each row will get its own scale.

(If block=None, it would use reshape(1, -1): one row holding all 8 weights, so one scale for everything.)

```python
s=np.abs(w).max(axis=1, keepdims=True) /qmax
```

```python
s
```

Checks whether any scale is 0. That only happens if a whole block is zeros, which would cause a divide-by-zero. Here nothing is 0, so nothing changes.

```python
s[s==0]=1.0
```

```python
q=np.clip(q, -qmax, qmax).astype(np.int8)
```

```python
q
```
