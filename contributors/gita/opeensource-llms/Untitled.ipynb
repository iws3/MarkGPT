```python
from gguf import GGUFReader
import glob, os
# ~/.cache/llama.cpp/*.gguf.

path = glob.glob(os.path.expanduser('~/llama.cpp/models/*.gguf'))[0]
r = GGUFReader(path)

for k, f in r.fields.items():
    if f.types and len(f.types) == 1 and f.types[0].name == 'STRING':
        print(k, '=', bytes(f.parts[f.data[0]]).decode()[:80])
    elif f.types and len(f.types) == 1:
        print(k, '=', f.parts[f.data[0]][0])

for t in r.tensors[:14]:
    print(t.name, list(t.shape), t.tensor_type.name)
```
