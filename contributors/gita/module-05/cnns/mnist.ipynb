## 1. Imports & data

## What is MNIST?

### MNIST stands for:

Modified National Institute of Standards and Technology Database

It is a collection of handwritten digits


#### 
Specifically:

- Digits 0 to 9
- Written by thousands of different people
- Converted into images
- Used to train computers to recognize handwriting

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
print("loaded..")
```

```python
# transforms: convert PIL image → tensor, then normalize pixel values
# mean=0.1307, std=0.3081 are the pre-computed MNIST dataset statistics
transform = transforms.Compose([
    transforms.ToTensor(),               
    transforms.Normalize((0.1307,), (0.3081,)) 
])
print("loaded")
transform
```

```python
transform
```

```python
train_data = datasets.MNIST(root='./data', train=True,
                             download=True, transform=transform)
test_data  = datasets.MNIST(root='./data', train=False,
                             download=True, transform=transform)
```

```python
train_data
```

```python

# DataLoader batches the data and shuffles each epoch
train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader  = DataLoader(test_data,  batch_size=64, shuffle=False)
print("hello loader")
```
