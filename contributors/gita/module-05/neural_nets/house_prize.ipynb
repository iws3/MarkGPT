```python
# preprocesing
housing.keys()
```

```python
X=housing.data.values.astype(np.float32)
y=housing.target.values.astype(np.float32)
```

```python
# Feature scaling — CRITICAL for ANNs
# Without scaling, features with large ranges dominate
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Convert to tensors
X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.float32).unsqueeze(1)  # [N] → [N, 1]
```

```python
print(X_tensor)
```

# Create dataset and split

```python

from torch.utils.data import DataLoader, TensorDataset, random_split
dataset=TensorDataset(X_tensor, y_tensor)
dataset
```

```python
train_size=int(0.7 * len(dataset))
val_size=int(0.15 * len(dataset))
test_size=len(dataset) - train_size-val_size
print(test_size)
print(val_size)
print(train_size)
```

# Create data loaders

```python
train_ds, val_ds, test_ds=random_split(dataset, [train_size, val_size, test_size])
train_loader=DataLoader(train_ds, batch_size=256, shuffle=True)
val_loader=DataLoader(val_ds, batch_size=256, shuffle=False)
test_loader=DataLoader(test_ds, batch_size=256, shuffle=False)

print(f"\n Train: {len(train_ds)} | val: {len(val_ds)} | Test: {len(test_ds)}")
```
