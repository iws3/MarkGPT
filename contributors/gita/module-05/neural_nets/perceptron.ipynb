```python
import torch
```

```python
# Imagine a simple neuron
input_features=3
output_features=1
```

```python
# lEARNABLE PARAMETERS
W=torch.randn(input_features, output_features, requires_grad=True)
b=torch.zeros(output_features, requires_grad=True)
print(b)
print(W)
W+b

```

```python
# One sample

x=torch.tensor([[1.0, 2.5, -1.0]])
x
```

```python
# Forward pass
z=x@W +b
y=torch.sigmoid(z)   # Activation
y
```

```python
# Backpropagation
loss=(y-0.8)**2
loss
```

```python
# loss - backpropagation
loss.backward()

print("Gradient of W: ", W.grad) 
print("Gradient of b:", b.grad)
```
