```python
import torch
import torch.nn as nn
```

**Understanding each component:**

`nn.Linear(in, out)` — The core transformation: `output = input @ W + b`

`nn.BatchNorm1d` — Normalizes each mini-batch to have mean≈0, std≈1. Stabilizes training dramatically. Think of it as re-centering the data after each layer.

`nn.ReLU()` — Activation function. Without it, stacking linear layers is mathematically equivalent to a single linear layer (no depth benefit).

`nn.Dropout(p)` — During training, randomly sets p% of neurons to zero each forward pass. Forces the network not to rely on any single neuron. Disabled during `model.eval()`.
